## 1. Why This Matters
Every dataset starts as a file somewhere -- a CSV export, a JSON API response, a plain
text log. Before pandas, before a database, someone has to open that file correctly.
This is literally the first thing Dana's 2am ingestion job (from the case study earlier
in this course) does: opens the raw export, reads it, hands it off. `brewbuzz_raw_orders.csv`,
sitting in `../data/`, *is* that raw export -- 100,150 rows, dirt and all.

## 2. Opening and Closing Files -- and Why `with` Exists

```python
f = open("../data/brewbuzz_raw_orders.csv", "r", encoding="utf-8")
first_line = f.readline()
f.close()
print(first_line.strip())
```

Forget to close a file, or close it too early, and the next read doesn't warn you --
it crashes:

```python
f.read()   # ValueError: I/O operation on closed file.
```

`with` closes the file automatically, even if an error happens partway through the
block. This is why every file-handling example from here on uses it.

```python
with open("../data/brewbuzz_raw_orders.csv", encoding="utf-8") as f:
    first_line = f.readline()
print(f.closed)   # True -- closed automatically the moment the block ends
```

## 3. Reading Line by Line
`.readline()` reads one line at a time. `.readlines()` reads *every* line into a list --
fine for a small file, wasteful for a 100,150-row one. Iterating the file object directly
is the memory-efficient middle ground: one line in memory at a time, no matter how big
the file is.

```python
with open("../data/brewbuzz_raw_orders.csv", encoding="utf-8") as f:
    header = f.readline()
    first_row = f.readline()
    second_row = f.readline()
print(header.strip())
print(first_row.strip())
print(second_row.strip())
```

```python
with open("../data/brewbuzz_raw_orders.csv", encoding="utf-8") as f:
    line_count = 0
    for line in f:          # one line in memory at a time, not all 100,150 at once
        line_count += 1
print(line_count)   # 100151 -- includes the header row
```

## 4. Writing and Appending Text Files
`"w"` mode overwrites the whole file -- if it already exists, its old contents are gone.
`"a"` mode appends to the end instead, without touching what's already there.

```python
with open("session_log.txt", "w", encoding="utf-8") as f:
    f.write("File handling session started\n")

with open("session_log.txt", "a", encoding="utf-8") as f:
    f.write("Loaded brewbuzz_raw_orders.csv\n")

with open("session_log.txt", encoding="utf-8") as f:
    print(f.read())
```

## 5. Reading CSV Properly -- the `csv` Module
You could split each line on commas yourself -- and it would quietly break the moment a
text field contains a comma (a product name like `"Latte, Large"` would look like two
columns instead of one). The `csv` module handles that correctly.

`csv.DictReader` also means every row is a dict keyed by column name --
`row["store_name"]`, not `row[2]` -- readable, and immune to the column order changing
later.

```python
import csv

with open("../data/brewbuzz_raw_orders.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    # print(reader.fieldnames)
    first_three = [next(reader) for _ in range(3)]


# print(first_three[3])
for row in first_three:
    print(row["order_id"], row["store_name"], row["product_name"], row["quantity"])
```

## 6. Writing CSV -- `csv.DictWriter`

```python

with open("../data/brewbuzz_raw_orders.csv", newline="", encoding="utf-8") as f_in, \
     open("airport_orders.csv", "w", newline="", encoding="utf-8") as f_out:
    reader = csv.DictReader(f_in)
    writer = csv.DictWriter(f_out, fieldnames=["order_id","order_date", "store_name", "customer_email", "order_total"])
    writer.writeheader()
         
    for row in reader:
        date = row["order_date"]
        if "2025-01-01" <= date <= "2025-12-31":   # exact match only, on purpose -- see step 8
            writer.writerow({"order_id": row["order_id"], "order_date":row["order_date"], "store_name": row["store_name"], "customer_email": row["customer_email"], "order_total": row["order_total"]})
            # print(row["store_name"])

with open("airport_orders.csv", encoding="utf-8") as f:
    print(sum(1 for _ in f) - 1, "rows written (excluding header)")   # 20016
```

## 7. Reading and Writing JSON
JSON is the standard format for structured data over the web -- nested objects and lists,
not just a flat grid like CSV. `json.dumps()` turns a Python object into a JSON string;
`json.dump()` writes it straight to a file. `json.loads()` / `json.load()` do the reverse.

```python
import json

summary = {
    "source_file": "brewbuzz_raw_orders.csv",
    "total_rows": 100150,
    "stores": ["BrewBuzz Airport", "BrewBuzz Downtown", "BrewBuzz Mall",
               "BrewBuzz Riverside", "BrewBuzz Uptown"],
}

with open("summary.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=4)

with open("summary.json", encoding="utf-8") as f:
    loaded = json.load(f)

print(loaded["total_rows"])
print(loaded["stores"][0])
```

## 8. Putting It Together
Read the raw file once, count rows per store using nothing but the `csv` module, and
write the result to both a text report and a JSON file -- the same two output formats a
real pipeline would produce, one for a human to skim (`.txt`) and one for another program
to consume (`.json`).

```python
counts = {}
with open("../data/brewbuzz_raw_orders.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    total_rows = 0
    for row in reader:
        total_rows += 1
        store = row["store_name"].strip()
        counts[store] = counts.get(store, 0) + 1

print(total_rows)   # 100150
for store, count in sorted(counts.items()):
    print(store, count)
```

Notice `.strip()` alone wasn't enough -- `'BrewBuzz Airport'` and `'brewbuzz airport'`
show up as two separate keys above, because the *casing* differs too, not just the
whitespace. That's exactly the kind of cleanup pandas handles far more comfortably, in
Week 4. File handling's job is to get the data **in** -- it doesn't have to get it
perfectly clean on the way.

```python
with open("store_counts.txt", "w", encoding="utf-8") as f:
    f.write(f"BrewBuzz raw orders -- {total_rows} total rows\n")
    f.write("-" * 40 + "\n")
    for store, count in sorted(counts.items(), key=lambda kv: -kv[1]):
        f.write(f"{store:<25} {count}\n")

with open("store_counts.json", "w", encoding="utf-8") as f:
    json.dump({"total_rows": total_rows, "counts_by_raw_label": counts}, f, indent=2)

with open("store_counts.txt", encoding="utf-8") as f:
    print(f.read())
```
