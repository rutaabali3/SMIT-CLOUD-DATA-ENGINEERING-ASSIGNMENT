## 2. File I/O Recap -- Two Ways In
Week 3 did this with the stdlib `csv` module: open the file, wrap it in a
`DictReader`, loop over rows. That still works -- it's just a lot of bookkeeping for
"give me a table."

```python
import csv

with open("../data/supermarket_sales.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for i, row in enumerate(reader):
        print(row["Branch"], row["City"], row["Total"])
        if i == 2:
            break
```

```python
import pandas as pd

df = pd.read_csv("../data/supermarket_sales.csv")
df.head(5)
```

One line replaced the loop, and `df.head(3)` already comes with column names and
inferred types attached -- bookkeeping `csv.DictReader` made you track by hand.

## 3. NumPy Refresher, On This Dataset
Week 4 Session 1 covered NumPy in depth on the BrewBuzz orders -- go back there for the
full lesson on why vectorized math beats a loop. Quick reminder here: every pandas
column already **is** a NumPy array underneath.

```python
import numpy as np

totals = df["Total"].to_numpy()
print(totals.dtype, totals.shape)
print("total revenue:", totals.sum().round(2))   # 322966.75
print("average sale: ", totals.mean().round(2))  # 322.97
print("largest sale: ", totals.max())            # 1042.65
```

```python
big_sales = totals > 500
print("sales over $500:", big_sales.sum())   # 227 out of 1000
```

`.to_numpy()` proves the point: `totals` is a plain NumPy array, so `totals > 500`
is the exact same boolean-mask trick from Week 4, not a new pandas concept.

## 4. Series vs DataFrame
A DataFrame is a dict of same-length Series sharing one row index. Pull one column out
and you get a `Series`; wrap the column name in a second pair of brackets and you get a
one-column `DataFrame` instead.

```python
print(type(df["Product line"]))       # Series
print(type(df[["Product line"]]))      # DataFrame
```

```python
print(df["Product line"])       # Series
print('------------------')
print(df[["Product line"]])      # DataFrame
```

## 6. Selecting and Filtering
`.loc` selects by label, `.iloc` by position, and boolean filtering (or the equivalent
`.query()`) picks rows by condition -- all three show up constantly in real analysis.

```python
df.loc[500:510, ["Branch", "City", "Total","gross margin percentage", "Product line"]]
```

```python
df.iloc[500:505, 10:15]
```

```python
yangon_members = df[(df["City"] == "Yangon") & (df["Customer type"] == "Member")]
print(len(yangon_members))   # boolean filter
```

```python
healthy_females = df[(df["Gender"] == "Female") & (df["Product line"] == "Health and beauty") & (df["Total"] > 500)]
print(len(healthy_females))
healthy_females.head()
```

```python
df_new = df[["City", "Total", "Gender", "Quantity"]]
print(len(df_new))
df_new.describe()
```

```python
df_loc = df.iloc[100:200, 5:15]
df_loc.tail()
```

```python
yangon_members_q = df.query('City == "Yangon" and `Customer type` == "Member"')
print(len(yangon_members_q))   # same result via .query() -- note the backticks for a
                                # column name with a space in it
yangon_members_q.head()
```

## 7. Sorting and Quick Aggregation

```python
df.sort_values("Total").head(5)[["Invoice ID", "Branch", "Total"]]
```

```python
df["Gender"].value_counts()
```

```python
df.groupby("City")["Total"].sum().round(3).sort_values()
```

## 8. Writing Data Back Out
Same DataFrame, three formats. `to_csv` for the universal format, `to_json` for
APIs/web use, `to_excel` for whoever downstream still opens Excel.

```python
top_branch_c = df[df["Branch"] == "C"]
top_branch_c.to_csv("../data/branch_c_sales.csv", index=False)
top_branch_c.to_json("../data/branch_c_sales.json", orient="records", indent=2)
top_branch_c.to_excel("../data/branch_c_sales.xlsx", index=False)
print(f"wrote {len(top_branch_c)} rows to csv/json/xlsx")
```

```python
check = pd.read_csv("../data/branch_c_sales.csv")
print(check.shape == top_branch_c.shape)   # True -- round trip preserved the data
```

Worth knowing: CSV has no type information. Re-read `Date` from that file and
it comes back as a plain string, not a datetime -- CSV remembers values, not dtypes.
Session 2 deals with re-parsing exactly that kind of column on purpose.
