## 2. From Lists to Arrays
A NumPy array looks like a list but isn't one: every element is the same fixed type,
stored contiguously in memory. That's the entire reason it's fast -- and the entire
reason it can't mix an `int` and a `str` in the same array the way a list can.

```python
import numpy as np

python_list = [1, 2, 3, 4, 5]
numpy_array = np.array(python_list)

print(type(python_list), type(numpy_array))
print(numpy_array.dtype)   # int64 -- every element is the same type
```

## 4. Cleaning the Raw Data Into Two Arrays
Reusing last session's exact skill: read the file, `try` to convert each row's
`quantity`/`unit_price`, and only keep the rows that actually parse *and* pass the
business rule (`> 0`). Error handling is what makes it safe to build these arrays from a
file you know has bad rows in it.

```python
import csv

quantities, prices = [], []
with open("../data/brewbuzz_raw_orders.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        try:
            q = int(row["quantity"])
            p = float(row["unit_price"])
            if q > 0 and p > 0:
                quantities.append(q)
                prices.append(p)
        except ValueError:
            continue   # already handled properly last session -- just skip here

qty_arr = np.array(quantities)
price_arr = np.array(prices)
print(len(qty_arr))   # 98866 -- clean rows kept out of 100,150
```

## 5. Vectorized Arithmetic -- Why It Beats a Loop

```python
import time

start = time.perf_counter()
totals_loop = [q * p for q, p in zip(quantities, prices)]
loop_time = time.perf_counter() - start

start = time.perf_counter()
totals_np = qty_arr * price_arr          # one line, no loop at all
np_time = time.perf_counter() - start

print(f"loop:  {loop_time*1000:.2f} ms")
print(f"numpy: {np_time*1000:.2f} ms")
print("same result?", np.allclose(totals_loop, totals_np))
```

On this machine that ran roughly 5.6ms for the loop versus 0.3ms for NumPy -- about
17x faster, on 98,866 rows, for one multiplication. NumPy pushes the entire loop down
into pre-compiled C code operating on contiguous memory, instead of Python evaluating one
`*` operation, one Python-object lookup, at a time. That gap only grows as the data does.

## 7. Boolean Masking & `np.where`
A boolean mask is an array of `True`/`False`, one per element -- indexing with it keeps
only the `True` positions. `np.where` builds a new array by choosing between two values
based on a condition, element by element, with no loop.

```python
mean, std = totals_np.mean(), totals_np.std()
print(f'mean is: {mean} and std is {std}')
outlier_mask = totals_np > mean + 3 * std
# print(totals_np > mean + 3 * std)
print(outlier_mask)
print("orders more than 3 std above the mean:", outlier_mask.sum())   # 2087
```

2,087 out of 98,866 (about 2%) is a lot more than a true bell curve would flag at 3
standard deviations (that would be well under 1%). That's a real limitation worth
knowing, not hiding: order totals here come from a handful of discrete product prices,
not a smooth continuous distribution, so a few popular high-price combinations repeat
often enough to look like "outliers" by this test when they're really just common. A
std-based cutoff assumes roughly-normal data -- always sanity-check that assumption
before trusting it.

```python
labels = np.where(totals_np < 5, "small", "regular")
print(labels)
print("small orders (<$5):", (labels == "regular").sum())   # 20677
```

## 8. Indexing & Slicing 2D Arrays
Same `[start:end]` rules from Week 2's string slicing, now with a second dimension:
`array[row_slice, column_slice]`.

```python
grid = np.arange(1, 13).reshape(3, 4)
print(grid)
```

```python
print(grid[2, 3])       # 7 -- row 1, column 2
print(grid[:2, 1:3])    # rows 0-1, columns 1-2
```
